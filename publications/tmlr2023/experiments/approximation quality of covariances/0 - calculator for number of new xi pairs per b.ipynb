```python
import numpy as np
import pandas as pd
```

```python
def get_increase(b):
    return 4 * b**3  - 6 * b**2 + 4*b - 1

b_domain = np.arange(1, 31)
increases = get_increase(b_domain)
pd.DataFrame({"b": b_domain, "increase": increases, "total": np.cumsum(increases)})
```

```python
def f1(b):
    return 4 * b**3 -6 *b**2 + 4*b-1

def f2(b):
    return 1 + 14*(b-1) + (b-1)*(b-2)/2 * 36 + (b-1)*(b-2)*(b-3)/6 * 24

b = 6
print(f1(b))
print(f2(b))
```
